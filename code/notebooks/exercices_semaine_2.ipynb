### Exercice - CALCUL DE LA DISTANCE ENTRE 2 POINTS

Créez un programme calculant et affichant la distance entre les 2 points, arrondie à 2 décimales.

La formule de la distance entre 2 points A(x₁, y₁) et B(x₂, y₂) est :
d = √((x₂ - x₁)² + (y₂ - y₁)²)

Représentez les deux points dans une seule liste `points` contenant deux sous-listes :
```python
points = [
    [50, 70],  # point A : [x1, y1]
    [80, 60],  # point B : [x2, y2]
]
```
Accédez aux coordonnées via les index : `points[1][0]` pour x₂. Pour réaliser la racine carrée de la formule, vous aurez besoin d'une fonction du module `math`.

✅ **Sortie attendue**
```
Point A
X1 : 50
Y1 : 70

Point B
X2 : 80
Y2 : 60

Distance : 31.62
```

````python
import math

points = [
    [50, 70],  # point A : [x1, y1]
    [80, 60],  # point B : [x2, y2]
]

x1 = points[0][0]
y1 = points[0][1]
x2 = points[1][0]
y2 = points[1][1]

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

print("Point A")
print(f"X1 : {x1}")
print(f"Y1 : {y1}")
print()
print("Point B")
print(f"X2 : {x2}")
print(f"Y2 : {y2}")
print()
print(f"Distance : {distance:.2f}")
````

### Exercice - CALCUL DE LA NOTE FINALE DU COURS

Créez un programme qui calcule la note finale obtenue par un étudiant dans ce cours. La règle de la double sanction n'est pas à appliquer dans cet exercice, car les notions nécessaires n'ont pas encore été vues.

**INSTRUCTIONS**
1. Déclarez une variable pour le nom de l'étudiant et quatre listes parallèles :
```python
nom_etudiant = "Nicola Tesla"
evaluations = ["TP1", "TP2", "Examen 1", "Examen 2", "Contrôles"]
notes = [8, 9, 90, 95, 9 ]
sur = [10, 10, 100, 100, 10 ]
ponderations = [0.15, 0.25, 0.20, 0.30, 0.10 ]
```
2. Appliquez les pondérations pour calculer la note finale en accédant aux éléments par leur index.
3. Affichez un résumé clair des notes de chaque évaluation et de la note finale, arrondies à 2 décimales.

✅ **Sortie attendue**
```
Étudiant : Nicola Tesla
TP1 (15%) : 8 / 10
TP2 (25%) : 9 / 10
Examen 1 (20%) : 90 / 100
Examen 2 (30%) : 95 / 100
Contrôles (10%) : 9 / 10

Note obtenue pour ce cours : 90.0 / 100
Attention, l'application de la double sanction n'a pas été réalisée!
```

````python
nom_etudiant = "Nicola Tesla"
evaluations = ["TP1", "TP2", "Examen 1", "Examen 2", "Contrôles"]
notes = [8, 9, 90, 95, 9 ]
sur = [10, 10, 100, 100, 10 ]
ponderations = [0.15, 0.25, 0.20, 0.30, 0.10 ]

note_tp1 = (notes[0] / sur[0]) * ponderations[0]
note_tp2 = (notes[1] / sur[1]) * ponderations[1]
note_ex1 = (notes[2] / sur[2]) * ponderations[2]
note_ex2 = (notes[3] / sur[3]) * ponderations[3]
note_ctrl = (notes[4] / sur[4]) * ponderations[4]

note_finale = (note_tp1 + note_tp2 + note_ex1 + note_ex2 + note_ctrl) * 100

print(f"Étudiant : {nom_etudiant}")
print(f"{evaluations[0]} ({ponderations[0]*100}%) : {notes[0]} / {sur[0]}")
print(f"{evaluations[1]} ({ponderations[1]*100}%) : {notes[1]} / {sur[1]}")
print(f"{evaluations[2]} ({ponderations[2]*100}%) : {notes[2]} / {sur[2]}")
print(f"{evaluations[3]} ({ponderations[3]*100}%) : {notes[3]} / {sur[3]}")
print(f"{evaluations[4]} ({ponderations[4]*100}%) : {notes[4]} / {sur[4]}")
print()
print(f"Note obtenue pour ce cours : {note_finale:.2f} / 100")
print("Attention, l'application de la double sanction n'a pas été réalisée!")
````

### Exercice - DÉCOMPOSER UN MONTANT EN BILLETS

Écrivez un programme permettant de décomposer un montant en billets de 100, 50, 20, 10 et 5$. Le programme doit privilégier les billets de plus grandes valeurs.

Voici un exemple d'exécution :
```
Montant : 376.54
Billets :
3 x 100$
1 x 50$
1 x 20$
0 x 10$
1 x 5$
Reste 1.54$
```

````python
montant_initial = 376.54
billets = [100, 50, 20, 10, 5] # coupures disponibles, du plus grand au plus petit

montant_restant = montant_initial

print(f"Montant : {montant_initial}")
print("Billets :")

# Billet de 100$
nb_billets_100 = montant_restant // billets[0]
montant_restant = montant_restant % billets[0]
print(f"{int(nb_billets_100)} x {billets[0]}$")

# Billet de 50$
nb_billets_50 = montant_restant // billets[1]
montant_restant = montant_restant % billets[1]
print(f"{int(nb_billets_50)} x {billets[1]}$")

# Billet de 20$
nb_billets_20 = montant_restant // billets[2]
montant_restant = montant_restant % billets[2]
print(f"{int(nb_billets_20)} x {billets[2]}$")

# Billet de 10$
nb_billets_10 = montant_restant // billets[3]
montant_restant = montant_restant % billets[3]
print(f"{int(nb_billets_10)} x {billets[3]}$")

# Billet de 5$
nb_billets_5 = montant_restant // billets[4]
montant_restant = montant_restant % billets[4]
print(f"{int(nb_billets_5)} x {billets[4]}$")

print(f"Reste {montant_restant:.2f}$")
````

### Exercice - Vitesse d'impact

Écrivez (sur papier) un programme qui calcule la vitesse d’impact (v) d’un objet en chute libre à partir d’une hauteur (h) donnée en mètres. (On néglige la résistance de l’air et on suppose que l’objet est lâché sans vitesse initiale.)

Formule : v = √(2 * g * h)

Note : utilisez g = 9.81 pour la constante de gravité.

**Contraintes techniques :**
- Déclarez une variable pour la hauteur.
- Importez le module nécessaire pour les calculs mathématiques.
- Arrondissez le résultat à 2 décimales.
- Affichez le résultat selon le format ci-dessous.

**Exemple attendu (pour h=5) :**
```
Hauteur : 5
La vitesse d'impact est de 9.90 m/s
```

````python
import math

h = 5  # en mètres
g = 9.81 # en m/s^2

v = math.sqrt(2 * g * h)

print(f"Hauteur : {h}")
print(f"La vitesse d'impact est de {v:.2f} m/s")
````
