Le modulo est très utile pour vérifier si un nombre est pair ou impair. Un nombre est pair si le reste de sa division par 2 est 0.

```python
10 % 2  # Est-ce que 10 est pair ?
```

```python
9 % 2   # Est-ce que 9 est pair ?
```

## Types de données

En Python, chaque valeur a un type. Les types de base sont les entiers (`int`), les nombres à virgule flottante (`float`), les chaînes de caractères (`str`) et les booléens (`bool`).

### Exercice 1: La fonction `type()`
Utilisez la fonction `type()` pour découvrir le type des valeurs suivantes.

```python
type(10)
```

```python
type(10.5)
```

```python
type("Bonjour")
```

```python
type(True)
```

### Exercice 2: Opérations selon le type
L'opérateur `+` se comporte différemment avec les nombres et avec les chaînes de caractères.

```python
# Addition de nombres
5 + 3
```

```python
# Concaténation de chaînes de caractères
"5" + "3"
```

### Exercice 3: Conversion de type (Casting)
On peut convertir une valeur d'un type à un autre. Par exemple, pour utiliser un nombre qui est sous forme de chaîne de caractères dans un calcul.

```python
nombre_en_texte = "25"
nombre_en_entier = int(nombre_en_texte)

nombre_en_entier + 10
```
