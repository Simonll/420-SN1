### Exercice - Jeu de devinette numérique

Écrivez un programme qui choisit un nombre entier aléatoire entre 0 et 100. L'utilisateur doit deviner ce nombre.

Tant que la réponse est incorrecte, le programme indique si le nombre à deviner est "Plus haut!" ou "Plus bas!".
Lorsqu'il trouve le bon nombre, le programme le félicite et s'arrête.

````python
import random

nombre_secret = random.randint(0, 100)
devine = -1 # Initialisation avec une valeur qui ne peut pas être la bonne

print("J'ai choisi un nombre entre 0 et 100. À vous de deviner!")

while devine != nombre_secret:
    devine = int(input("Votre proposition : "))
    
    if devine < nombre_secret:
        print("Plus haut!")
    elif devine > nombre_secret:
        print("Plus bas!")

print(f"Bravo! Le nombre était bien {nombre_secret}.")
````

### Exercice - Calculer des statistiques (sans fonctions natives)

Soit une liste de températures. Calculez et affichez le nombre de mesures, la somme, la moyenne, le minimum et le maximum **sans utiliser** les fonctions `len()`, `sum()`, `min()` et `max()`.

Vous devrez utiliser une boucle et des variables accumulatrices.

````python
temperatures = [23.5, 25.4, 27.6, 21.9, 26.7, 24.6]

if not temperatures: # Gérer le cas d'une liste vide
    print("Aucune donnée à analyser.")
else:
    # Initialisation des accumulateurs
    nb_mesures = 0
    somme = 0
    # Pour min/max, on initialise avec la première valeur de la liste
    temp_min = temperatures[0]
    temp_max = temperatures[0]

    # Boucle pour parcourir toutes les températures
    for temp in temperatures:
        nb_mesures += 1
        somme += temp
        
        if temp < temp_min:
            temp_min = temp
        if temp > temp_max:
            temp_max = temp
            
    moyenne = somme / nb_mesures

    # Affichage des résultats
    print(f"Températures : {temperatures}")
    print(f"Nombre de mesures : {nb_mesures}")
    print(f"Somme : {somme}")
    print(f"Moyenne : {moyenne:.2f}")
    print(f"Minimum : {temp_min}")
    print(f"Maximum : {temp_max}")
````

### Exercice - Affichage d'étoiles (pyramide)

Écrivez une fonction `pyramide(hauteur)` qui affiche une pyramide d'étoiles de la hauteur donnée.

**Exemple pour `pyramide(4)` :**
```
   *
  ***
 *****
*******
```
**Indices :**
- Vous aurez besoin de boucles imbriquées.
- La boucle extérieure gère les lignes (de 1 à `hauteur`).
- La boucle intérieure (ou une multiplication de chaînes) gère l'affichage des espaces et des étoiles pour chaque ligne.
- Pour une ligne `i`, il faut `hauteur - i` espaces et `2 * i - 1` étoiles.

````python
def pyramide(hauteur):
    """Affiche une pyramide d'étoiles de la hauteur spécifiée."""
    for i in range(1, hauteur + 1):
        # Calculer le nombre d'espaces et d'étoiles
        espaces = hauteur - i
        etoiles = 2 * i - 1
        
        # Construire la ligne et l'afficher
        ligne = ' ' * espaces + '*' * etoiles
        print(ligne)

# Test
pyramide(4)
````

# L'annulation catastrophique et la variance

En statistiques, on apprend souvent cette formule rapide pour calculer la variance d'une population :

$$Var = \frac{\sum x^2}{n} - \left(\frac{\sum x}{n}\right)^2$$

Sur papier, elle est géniale. Mais pour un ordinateur, c'est une bombe à retardement. Si vos données sont de grands nombres très proches les uns des autres, l'ordinateur perd sa précision lors de la soustraction. C'est ce qu'on appelle l'**annulation catastrophique**.

````python
def variance_naive(donnees):
    """Calcule la variance en utilisant la formule théorique (naïve)."""
    n = len(donnees)
    if n < 2:
        return float('nan')
    
    somme = sum(donnees)
    somme_carres = sum(x**2 for x in donnees)
    
    # Formule : E[X^2] - (E[X])^2
    variance = (somme_carres / n) - (somme / n)**2
    return variance
````

## La solution : L'algorithme de Welford

L'algorithme de Welford (1962) est la solution robuste. Plutôt que d'additionner des carrés qui deviennent gigantesques, il met à jour la moyenne et la somme des carrés des écarts de façon incrémentale, un point de données à la fois.

C'est ce qu'utilisent les vraies bibliothèques scientifiques.

````python
def variance_welford(donnees):
    """Calcule la variance en utilisant l'algorithme robuste de Welford."""
    compteur = 0
    moyenne = 0.0
    M2 = 0.0  # Somme des carrés des écarts à la moyenne

    for x in donnees:
        compteur += 1
        
        # Écart avec l'ancienne moyenne
        delta = x - moyenne
        
        # Mise à jour de la moyenne
        moyenne += delta / compteur
        
        # Écart avec la NOUVELLE moyenne
        delta2 = x - moyenne
        
        # Mise à jour de la somme des carrés des écarts
        M2 += delta * delta2

    if compteur < 2:
        return float('nan')

    variance_echantillon = M2 / (compteur - 1)
    variance_population = M2 / compteur

    return variance_echantillon, variance_population
````

## Le Crash-Test
Voici ce qui se passe si on teste un ensemble de données conçu spécifiquement pour faire "planter" la formule statistique naïve. Nous avons 3 nombres identiques, la variance **doit** donc être de `0`.

````python
donnees_extremes = [1000000000.0, 1000000000.0, 1000000000.0]

print("--- Test avec des nombres extrêmes ---")

# 1. Test avec la formule naïve
var_naive = variance_naive(donnees_extremes)
print(f"Formule naïve : {var_naive}")
if var_naive < 0:
    print("ERREUR : La variance est négative, c'est mathématiquement impossible !\n")

# 2. Test avec l'algorithme de Welford
var_ech, var_pop = variance_welford(donnees_extremes)
print(f"Algorithme de Welford (Pop) : {var_pop}")
if var_pop == 0.0:
    print("✅ SUCCÈS : La variance est bien calculée à 0.0")
````
