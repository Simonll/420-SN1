## 1. Expressions booléennes

Les expressions booléennes retournent `True` ou `False`.

```python
# Expressions simples
5 > 3      # True
5 < 3      # False
5 == 5     # True
5 != 3     # True
```

## 2. Conditions (if/elif/else)

```python
# Exemple simple avec if/elif/else
concentration = 50
seuil_min = 20
seuil_max = 80

if concentration < seuil_min:
    print("INSUFFISANT")
elif concentration <= seuil_max:
    print("OPTIMAL")
else:
    print("TOXIQUE / CRITIQUE")
```

## 3. Comparaisons

Attention aux cas limites avec `==` et `<=`

```python
# Cas limites
concentration = 20
seuil_min = 20
seuil_max = 80

# Tester les seuils
print(f"concentration < seuil_min: {concentration < seuil_min}")  # False
print(f"concentration <= seuil_max: {concentration <= seuil_max}")  # True
print(f"concentration == seuil_min: {concentration == seuil_min}")  # True
```

## 4. Problèmes sans validation

Que se passe-t-il si on ne valide pas les paramètres?

```python
# Sans validation - résultat absurde
def densite(masse, volume):
    return masse / volume

print(densite(10, 2))    # 5.0 OK
print(densite(-3, 5))    # -0.6 Absurde! Pas d'avertissement
# print(densite(8, 0))   # ZeroDivisionError - programme plante
```

## 5. Validation avec `raise ValueError()`

Utiliser `raise` pour contrôler les erreurs avec des messages clairs.

```python
def densite_validee(masse, volume):
    if masse < 0:
        raise ValueError("La masse doit être positive.")
    if volume <= 0:
        raise ValueError("Le volume doit être supérieur à 0.")
    return masse / volume

# Cas valides
print(densite_validee(10, 2))    # OK: 5.0

# Cas invalides - décommentez pour voir l'erreur
# print(densite_validee(-3, 5))    # ValueError: La masse doit être positive.
# print(densite_validee(8, 0))     # ValueError: Le volume doit être supérieur à 0.
```

## 6. Vérification de type avec `isinstance()`

Vérifier que les paramètres sont du bon type avant de les utiliser.

```python
# Vérifier les types
print(isinstance(5, int))           # True
print(isinstance(5.0, float))       # True
print(isinstance(5, float))         # False
print(isinstance("5", int))         # False

# Vérifier plusieurs types à la fois
print(isinstance(5, (int, float)))      # True
print(isinstance(5.0, (int, float)))    # True
print(isinstance("5", (int, float)))    # False
```

## 7. Validation complète: type puis valeur

Toujours vérifier le TYPE en premier, puis les VALEURS.

```python
def calculer_concentration(masse_g, volume_L):
    # Vérifier les types
    if not isinstance(masse_g, (int, float)):
        raise TypeError("La masse doit être un nombre (int ou float).")
    if not isinstance(volume_L, (int, float)):
        raise TypeError("Le volume doit être un nombre (int ou float).")
    
    # Vérifier les valeurs
    if masse_g < 0:
        raise ValueError("La masse doit être positive.")
    if volume_L <= 0:
        raise ValueError("Le volume doit être supérieur à 0.")
    
    return masse_g / volume_L

# Cas valides
print(calculer_concentration(10, 2))      # OK: 5.0

# Cas invalides - décommentez pour voir l'erreur
# print(calculer_concentration(-5, 2))     # ValueError: masse négative
# print(calculer_concentration(10, 0))     # ValueError: volume nul
# print(calculer_concentration(10, "a"))   # TypeError: volume pas un nombre
```

## 8. Gestion des erreurs avec `try/except`

Attraper les erreurs pour empêcher le programme de planter.

```python
# Sans try/except - le programme s'arrête à la première erreur
# print(calculer_concentration(10, 2))      # OK
# print(calculer_concentration(-5, 2))      # ERREUR - programme s'arrête ici
# print(calculer_concentration(10, 0))      # Jamais atteint

# Avec try/except - le programme continue
try:
    print(calculer_concentration(10, 2))      # OK
    print(calculer_concentration(-5, 2))      # ERREUR - on va au except
    print(calculer_concentration(10, 0))      # Jamais atteint
except (ValueError, TypeError) as e:
    print(f"Erreur interceptée: {e}")

# Le programme continue après le try/except
print("Programme continue normalement")
```

## 9. Exemple complet: evaluer_niveau()

Fonction avec validation complète, comme dans le TP1.

```python
def evaluer_niveau(concentration, seuil_min, seuil_max):
    """Évalue le niveau de concentration.
    
    Paramètres:
    concentration (float): valeur à évaluer
    seuil_min (float): seuil minimum
    seuil_max (float): seuil maximum
    
    Retourne:
    str: "INSUFFISANT", "OPTIMAL", ou "TOXIQUE / CRITIQUE"
    """
    # Validation des types
    if not isinstance(concentration, (int, float)):
        raise TypeError("concentration doit être un nombre")
    if not isinstance(seuil_min, (int, float)):
        raise TypeError("seuil_min doit être un nombre")
    if not isinstance(seuil_max, (int, float)):
        raise TypeError("seuil_max doit être un nombre")
    
    # Validation des valeurs
    if seuil_min >= seuil_max:
        raise ValueError("seuil_min doit être < seuil_max")
    
    # Logique
    if concentration < seuil_min:
        return "INSUFFISANT"
    elif concentration <= seuil_max:
        return "OPTIMAL"
    else:
        return "TOXIQUE / CRITIQUE"

# Tests
print(evaluer_niveau(15, 20, 80))  # INSUFFISANT
print(evaluer_niveau(20, 20, 80))  # OPTIMAL (cas limite)
print(evaluer_niveau(50, 20, 80))  # OPTIMAL
print(evaluer_niveau(80, 20, 80))  # OPTIMAL (cas limite)
print(evaluer_niveau(85, 20, 80))  # TOXIQUE / CRITIQUE
```

## 10. Bonnes pratiques

- Vérifier d'abord les TYPES, puis les VALEURS
- Utiliser des messages clairs dans `raise`
- N'attraper que les erreurs PRÉVUES avec `except`
- Toujours documenter les exceptions possibles dans la docstring
