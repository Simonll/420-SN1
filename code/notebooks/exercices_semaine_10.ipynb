### Exercice - Écrire et lire un fichier\n\nÉcrivez un script qui :\n1. Crée une liste de chaînes de caractères.\n2. Écrit chaque chaîne sur une nouvelle ligne dans un fichier nommé `mon_fichier.txt`.\n3. Relit le fichier et affiche son contenu à la console.

```python
lignes_a_ecrire = [
    "Ceci est la première ligne.",
    "Voici la deuxième.",
    "Et enfin, la troisième."
]

# Écriture dans le fichier
with open("mon_fichier.txt", "w", encoding="utf-8") as f:
    for ligne in lignes_a_ecrire:
        f.write(ligne + "\n")

print("Fichier 'mon_fichier.txt' créé.")

# Lecture et affichage du fichier
print("\nContenu du fichier :")
with open("mon_fichier.txt", "r", encoding="utf-8") as f:
    contenu = f.read()
    print(contenu)
```

### Exercice - Opérations vectorielles avec NumPy\n\nUtilisez NumPy pour effectuer des calculs sur un tableau de températures en Celsius et les convertir en Fahrenheit.\n\n**Formule :** °F = (°C × 9/5) + 32

```python
import numpy as np

temperatures_c = np.array([0, 10, 20, 30, 100])

# Conversion vectorielle (sans boucle)
temperatures_f = (temperatures_c * 9/5) + 32

print(f"Températures en Celsius : {temperatures_c}")
print(f"Températures en Fahrenheit : {temperatures_f}")
```

### Exercice - Statistiques avec NumPy\n\nÀ partir d'un tableau NumPy de notes, calculez la moyenne, le minimum, le maximum et le nombre de notes au-dessus de 60.

```python
import numpy as np

notes = np.array([85, 92, 58, 74, 60, 99, 45, 78])

moyenne = np.mean(notes)
note_min = np.min(notes)
note_max = np.max(notes)

# Filtrage booléen pour compter les réussites
reussites = notes[notes >= 60]
nb_reussites = len(reussites)

print(f"Notes : {notes}")
print(f"Moyenne : {moyenne:.2f}")
print(f"Note minimale : {note_min}")
print(f"Note maximale : {note_max}")
print(f"Nombre de notes >= 60 : {nb_reussites}")
```
