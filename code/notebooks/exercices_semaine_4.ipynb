### Exercice - Le niveau de pH

Définissez une fonction `niveau_ph(ph)` qui reçoit une valeur de pH et retourne une chaîne de caractères indiquant si la solution est "acide", "neutre" ou "basique".
- pH < 7 : acide
- pH = 7 : neutre
- pH > 7 : basique

Testez votre fonction avec les valeurs 5, 7 et 10.

```python
def niveau_ph(ph):
    """Retourne la classification d'une solution selon son pH."""
    if ph < 7:
        return "acide"
    elif ph == 7:
        return "neutre"
    else:
        return "basique"

# Tests
print(f"Une solution avec un pH de 5 est : {niveau_ph(5)}")
print(f"Une solution avec un pH de 7 est : {niveau_ph(7)}")
print(f"Une solution avec un pH de 10 est : {niveau_ph(10)}")
```

### Exercice - Refactorisation : Bulletins scolaires

Le code ci-dessous affiche un bulletin pour un étudiant. Il fonctionne, mais il est très répétitif. Votre mission est de le réécrire proprement en appliquant le principe DRY (Don't Repeat Yourself).

**Instructions :**
1. Identifiez les instructions `print` qui sont communes à tous les blocs.
2. Créez une fonction `afficher_bulletin(nom, note)` qui réorganise le code pour éviter la répétition.
3. Les seules instructions qui devraient rester à l'intérieur des conditions sont celles qui changent (la mention et le message final).
4. Appelez votre nouvelle fonction pour valider qu'elle produit le même résultat.

```python
def afficher_bulletin(nom, note):
    """Affiche un bulletin scolaire formaté pour un étudiant."""
    
    # --- Partie commune (avant) ---
    print("=== Bulletin scolaire ===")
    print(f"Étudiant : {nom}")
    print(f"Note : {note} %")

    # --- Partie conditionnelle ---
    if note >= 90:
        mention = "Excellent"
        message = "Félicitations!"
    elif note >= 75:
        mention = "Bien"
        message = "Félicitations!"
    elif note >= 60:
        mention = "Passable"
        message = "Bonne continuation."
    else:
        mention = "Échec"
        message = "Bonne continuation."
    
    print(f"Mention : {mention}")
    print(message)

    # --- Partie commune (après) ---
    print("=========================")
    print() # Ligne vide pour séparer les bulletins

# Tests
afficher_bulletin("Félix", 72)
afficher_bulletin("Dayron", 91)
```

### Exercice - Tarification électrique par paliers

Hydro-Québec facture l'électricité selon un système de paliers. Voici une version simplifiée :
- **Tarif D (régulier)** : 0,06 $/kWh pour les 40 premiers kWh/jour, puis 0,10 $/kWh pour le reste.
- **Tarif nuit (uniforme)** : 0,05 $/kWh pour toute la consommation.

**Instructions :**
1. Créez une fonction `calculer_facture(kwh, tarif_nuit=False)` qui calcule et retourne le montant à payer.
2. Si `tarif_nuit` est `True`, le calcul est simple. Sinon, vous devrez utiliser un `if` pour gérer les deux paliers du tarif D.
3. Testez votre fonction avec les appels fournis.

```python
def calculer_facture(kwh, tarif_nuit=False):
    """Calcule la facture d'électricité selon le tarif et la consommation."""
    if tarif_nuit:
        return kwh * 0.05
    else: # Tarif D
        if kwh <= 40:
            return kwh * 0.06
        else:
            cout_palier1 = 40 * 0.06
            cout_palier2 = (kwh - 40) * 0.10
            return cout_palier1 + cout_palier2

# Tests
print(f"Facture pour 30 kWh (Tarif D): {calculer_facture(30):.2f} $")
print(f"Facture pour 65 kWh (Tarif D): {calculer_facture(65):.2f} $")
print(f"Facture pour 65 kWh (Tarif Nuit): {calculer_facture(65, True):.2f} $")
```

### Exercice - Indice de masse corporelle (IMC)

Écrivez un programme qui calcule et interprète l'IMC d'une personne.

**Formule :** IMC = masse / taille²

**Interprétation :**
| IMC | Catégorie |
|---|---|
| < 18.5 | Insuffisance pondérale |
| < 25.0 | Poids normal |
| < 30.0 | Surpoids |
| ≥ 30.0 | Obésité |

**Contraintes :**
- Définissez une fonction `calculer_imc(masse, taille)` qui retourne l'IMC arrondi à 1 décimale.
- Définissez une fonction `interpreter_imc(imc)` qui retourne la catégorie correspondante.
- Affichez le résultat final de manière claire.

```python
def calculer_imc(masse, taille):
    """Calcule l'IMC à partir de la masse (kg) et la taille (m)."""
    if taille == 0:
        return 0
    imc = masse / (taille ** 2)
    return round(imc, 1)

def interpreter_imc(imc):
    """Retourne la catégorie de poids selon l'IMC."""
    if imc < 18.5:
        return "Insuffisance pondérale"
    elif imc < 25.0:
        return "Poids normal"
    elif imc < 30.0:
        return "Surpoids"
    else:
        return "Obésité"

# Données
masse_kg = 70
taille_m = 1.75

# Calculs
imc_calcule = calculer_imc(masse_kg, taille_m)
categorie = interpreter_imc(imc_calcule)

# Affichage
print(f"Masse : {masse_kg} kg")
print(f"Taille : {taille_m} m")
print(f"IMC : {imc_calcule}")
print(f"Catégorie : {categorie}")
```
