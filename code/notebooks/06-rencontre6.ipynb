### Exercice 1 - Partie entière d'une racine

Avant de trouver les décimales d'une racine carrée, il faut trouver sa partie entière. C'est le plus grand entier `i` tel que `i² <= n`.

Écrivez une fonction `partie_entiere_racine(n)` qui trouve et retourne cette partie entière en testant les entiers 0, 1, 2, 3, ... jusqu'à trouver la bonne valeur.

```python
def partie_entiere_racine(n):
    """Trouve la partie entière de la racine carrée de n."""
    if n < 0:
        return None # Ou lever une exception
    
    i = 0
    while (i + 1) ** 2 <= n:
        i += 1
    return i

# Tests
print(f"Partie entière de la racine de 88 : {partie_entiere_racine(88)}") # 9
print(f"Partie entière de la racine de 9 : {partie_entiere_racine(9)}")   # 3
print(f"Partie entière de la racine de 2 : {partie_entiere_racine(2)}")   # 1
```

### Exercice 2 - Méthode par dichotomie (simplifiée)

La dichotomie (ou bissection) est une méthode pour trouver une solution dans un intervalle en le coupant en deux à chaque étape.

Écrivez une fonction `trouver_racine_dichotomie(n, tolerance=0.001)` qui trouve la racine carrée de `n`.

**Logique :**
1. Initialisez un intervalle `[bas, haut]` (par exemple, `[0, n]`).
2. Tant que la largeur de l'intervalle (`haut - bas`) est plus grande que la `tolerance` :
   a. Calculez le `milieu`.
   b. Si `milieu² < n`, la racine est dans la moitié supérieure, donc `bas = milieu`.
   c. Sinon, la racine est dans la moitié inférieure, donc `haut = milieu`.
3. Retournez la borne `haut` (ou `bas`, elles seront très proches).

```python
def trouver_racine_dichotomie(n, tolerance=0.0001):
    """Trouve la racine carrée de n par dichotomie."""
    if n < 0:
        return None
    
    bas, haut = 0, n
    # Cas particulier pour les nombres entre 0 et 1
    if n < 1:
        haut = 1

    while (haut - bas) > tolerance:
        milieu = (bas + haut) / 2
        if milieu ** 2 < n:
            bas = milieu
        else:
            haut = milieu
    return haut

# Test
import math
racine_2 = trouver_racine_dichotomie(2)
print(f"Racine de 2 (dichotomie) : {racine_2}")
print(f"Racine de 2 (math.sqrt)   : {math.sqrt(2)}")
```

### Exercice 3 - Construction chiffre par chiffre

L'autre méthode du TP consiste à construire la racine chiffre par chiffre. Cet exercice simplifie la logique.

Écrivez une fonction `construire_nombre(chiffres)` qui prend une liste de chiffres (ex: `[2, 8, 2]`) et retourne le nombre décimal correspondant (ex: `2.82`).

**Indice :**
Parcourez la liste de chiffres. Pour chaque chiffre, ajoutez-le au nombre en cours en le divisant par la bonne puissance de 10 (10⁰, 10¹, 10², etc.).

```python
def construire_nombre(chiffres):
    """Construit un nombre décimal à partir d'une liste de chiffres."""
    if not chiffres:
        return 0
    
    nombre = chiffres[0] # La partie entière
    
    # La partie décimale
    for i in range(1, len(chiffres)):
        nombre += chiffres[i] * (10 ** -i)
        
    return nombre

# Test
chiffres_racine = [2, 8, 2, 8, 4]
resultat = construire_nombre(chiffres_racine)
print(f"Le nombre construit à partir de {chiffres_racine} est : {resultat}")
```

### Exercice 4 - Mesure de performance

Le TP demande de comparer la performance de vos fonctions. Le module `time` est parfait pour ça.

Écrivez un script qui :
1. Enregistre le temps **avant** d'appeler une fonction (ex: `math.sqrt`) dans une boucle qui s'exécute 100 000 fois.
2. Enregistre le temps **après** la boucle.
3. Calcule et affiche la durée totale en millisecondes.

```python
import time
import math

nb_iterations = 100_000

# Mesure pour math.sqrt
start_time = time.monotonic_ns()

for i in range(nb_iterations):
    math.sqrt(i) # On appelle la fonction à mesurer

end_time = time.monotonic_ns()

duree_ns = end_time - start_time
duree_ms = duree_ns / 1_000_000 # Conversion de nanosecondes en millisecondes
temps_moyen_ms = duree_ms / nb_iterations

print(f"Temps total pour {nb_iterations} appels : {duree_ms:.2f} ms")
print(f"Temps moyen par appel : {temps_moyen_ms:.6f} ms")
```

### Exercice 5 - Généraliser à la racine n-ième

Le TP demande de généraliser vos fonctions pour calculer la racine n-ième (carrée, cubique, etc.). Cela se fait en remplaçant l'exposant `2` par un paramètre `base`.

Modifiez votre fonction `partie_entiere_racine` pour qu'elle accepte un paramètre `base` (avec une valeur par défaut de 2).

```python
def partie_entiere_racine_nieme(n, base=2):
    """Trouve la partie entière de la racine n-ième de n."""
    if n < 0 and base % 2 == 0: # Racine paire d'un négatif
        raise ValueError("Racine paire d'un nombre négatif non définie dans les réels.")
    
    # Gérer les négatifs pour les bases impaires
    if n < 0:
        # On cherche la racine de la valeur absolue et on la rend négative
        return -partie_entiere_racine_nieme(abs(n), base)
    
    i = 0
    while (i + 1) ** base <= n:
        i += 1
    return i

# Tests
print(f"Partie entière de la racine CUBIQUE de 27 : {partie_entiere_racine_nieme(27, 3)}") # 3
print(f"Partie entière de la racine CUBIQUE de -27 : {partie_entiere_racine_nieme(-27, 3)}")# -3
print(f"Partie entière de la racine QUATRIÈME de 80 : {partie_entiere_racine_nieme(80, 4)}")# 2
```
