### Question 3 - Écrire du code pour résoudre un problème

**Sans module ni bibliothèque.**

**3a.** Écrivez une fonction `frequences(phrase)` qui retourne un dictionnaire `mot -> nombre d'apparitions`. Les mots sont séparés par des espaces et la comparaison ne tient pas compte des majuscules.

**3b.** Écrivez une fonction `bulletin(etudiants)` qui reçoit une liste de tuples `(nom, [notes])` et retourne une liste de tuples `(nom, moyenne, "réussite" ou "échec")`. La réussite est une moyenne de 60 ou plus. N'utilisez ni `sum()`, ni `max()`, ni `min()`.

````python
def frequences(phrase):
    compte = {}
    for mot in phrase.lower().split():
        if mot in compte:
            compte[mot] += 1
        else:
            compte[mot] = 1
    return compte


print(frequences("Le chat voit le chien et le chien voit le chat"))
````

````python
def bulletin(etudiants):
    resultat = []
    for nom, notes in etudiants:
        total = 0
        for note in notes:
            total += note
        moyenne = total / len(notes)
        statut = "réussite" if moyenne >= 60 else "échec"
        resultat.append((nom, round(moyenne, 1), statut))
    return resultat


groupe = [("Léa", [72, 85, 64]), ("Mathis", [48, 61, 55]), ("Yasmine", [90, 88, 95])]
for nom, moyenne, statut in bulletin(groupe):
    print(f"{nom:<8} {moyenne:>5} {statut}")
````

1. Tracez **une courbe par ville** (température en fonction du jour) sur le même graphique, avec une légende ;

Utilisez le fichier `Turing.csv` (colonnes `year`, `laureate`, `contribution`, `affiliation`) et répondez aux questions suivantes :

2. Quelles sont les 3 affiliations ayant le plus de lauréats ?
3. Quels lauréats ont une contribution contenant le mot « programming » (sans tenir compte des majuscules) ?
4. Combien de lauréats ont été récompensés avant 2000 ? Et à partir de 2000 ?
5. Combien d'années distinctes figurent dans le fichier, et quelles années comptent plus d'un lauréat ?
6. Affichez, triés par ordre alphabétique, les lauréats affiliés à une université (le nom de l'affiliation contient « University »).

### Question 6 - Synthèse

Une association environnementale veut suivre l'évolution de la consommation des voitures du fichier `mpg_dataset.csv`.

1. Chargez les données, retirez les lignes où `horsepower` est manquant et supprimez les éventuels doublons ;
2. Ajoutez une colonne `annee` = `1900 + model_year` et une colonne `l_par_100km` = `235.215 / mpg` (arrondie à 1 décimale) ;
3. Avec `apply()`, ajoutez une colonne `classe` : « économique » si `l_par_100km < 9`, « moyenne » si elle est inférieure à 13, « énergivore » sinon ;
4. Affichez le nombre de voitures par classe, puis la consommation moyenne par origine ;
5. Calculez la consommation moyenne par année **pour chaque origine** et tracez, avec Matplotlib (sans les graphiques de Pandas), une courbe par origine sur le même graphique ;
6. Exportez les voitures « économiques » dans `voitures_economiques.csv` (sans l'index) et vérifiez l'export en le relisant.

````python
voitures = pd.read_csv(chemin_donnees("mpg_dataset.csv"))
voitures = voitures.dropna(subset=["horsepower"]).drop_duplicates()

voitures["annee"] = 1900 + voitures["model_year"]
voitures["l_par_100km"] = (235.215 / voitures["mpg"]).round(1)


def classer(consommation):
    if consommation < 9:
        return "économique"
    elif consommation < 13:
        return "moyenne"
    else:
        return "énergivore"


voitures["classe"] = voitures["l_par_100km"].apply(classer)

print("Voitures par classe :")
print(voitures.groupby("classe")["name"].count())
print("\nConsommation moyenne par origine (L/100 km) :")
print(voitures.groupby("origin")["l_par_100km"].mean().round(2))

for origine in ["usa", "europe", "japan"]:
    sous_ensemble = voitures[voitures["origin"] == origine]
    moyenne_par_annee = sous_ensemble.groupby("annee")["l_par_100km"].mean()
    plt.plot(list(moyenne_par_annee.index), list(moyenne_par_annee.values), "o-", label=origine)
plt.title("Consommation moyenne selon l'année et l'origine")
plt.xlabel("Année du modèle")
plt.ylabel("Consommation (L/100 km)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

economiques = voitures[voitures["classe"] == "économique"]
economiques.to_csv("voitures_economiques.csv", index=False)
relu = pd.read_csv("voitures_economiques.csv")
print("\nVoitures économiques exportées :", len(relu))
print(relu[["name", "annee", "l_par_100km"]].head())
````
