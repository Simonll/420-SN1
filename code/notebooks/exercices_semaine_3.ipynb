### Exercice - Portée des variables

Soit le code suivant :
```python
def calcul_complexe(a, b, c):
    x = a ** b
    y = c * (b - a)
    return x + y

x = 5
calcul_complexe(2, 3, 4)
print(x)
```
1. Combien vaut `x` dans `calcul_complexe`? (ligne 2)
2. Combien vaut `x` dans le script principal? (ligne 6)
3. S’agit‑il de la même variable `x` ou de deux variables distinctes? Expliquez.

#### Solution
1.  **Dans `calcul_complexe`, `x` vaut 8.** Le `x` à la ligne 2 est une variable **locale** à la fonction. Sa valeur est `a ** b`, soit `2 ** 3 = 8`.
2.  **Dans le script principal, `x` vaut 5.** Le `x` à la ligne 6 est une variable **globale**. Sa valeur est initialisée à 5 et n'est jamais modifiée par la suite.
3.  Il s'agit de **deux variables distinctes**. La variable `x` locale (dans la fonction) "masque" la variable globale `x` (à l'extérieur) mais ne la modifie pas. La portée de la variable locale est limitée à l'intérieur de la fonction `calcul_complexe`.

### Exercice - FORCE GRAVITATIONNELLE

Dans le même module créé à l'exercice précédent, ajoutez une fonction permettant de calculer la force gravitationnelle en utilisant la loi de la gravitation universelle de Newton dont la formule est :
F = (G × m1 × m2) / r²
où :
- G = 6.674 × 10⁻¹¹ (constante gravitationnelle),
- m1 et m2 sont les masses en kg,
- r est la distance en mètres.

````python
# Cette cellule simule l'ajout de la fonction au fichier 'sciences_nature.py'
G = 6.674e-11

def force_gravitationnelle(m1, m2, r):
    """Calcule la force gravitationnelle entre deux objets."""
    if r == 0:
        return float('inf') # La force est infinie si la distance est nulle
    return (G * m1 * m2) / (r ** 2)

# Test avec la Terre et la Lune
masse_terre = 5.972e24  # kg
masse_lune = 7.342e22   # kg
distance_terre_lune = 3.844e8 # mètres

force = force_gravitationnelle(masse_terre, masse_lune, distance_terre_lune)
print(f"Force gravitationnelle entre la Terre et la Lune : {force:.2e} N")
````

### Exercice - Refactorisation du code - Calcul de densité

Réécrivez le programme ci-dessous en utilisant des listes parallèles et une fonction pour éviter la répétition du code.

````python
# --- Données initiales ---
masses_g = [135, 270, 92]
volumes_cm3 = [84, 151, 113]

# --- Fonction de calcul ---
def calculer_densite(masse_g, volume_cm3):
    """Calcule la densité en kg/m³ à partir de g et cm³."""
    masse_kg = masse_g / 1000
    volume_m3 = volume_cm3 / 1_000_000
    densite = masse_kg / volume_m3
    return densite

# --- Script principal ---
densite1 = calculer_densite(masses_g[0], volumes_cm3[0])
densite2 = calculer_densite(masses_g[1], volumes_cm3[1])
densite3 = calculer_densite(masses_g[2], volumes_cm3[2])

# --- Affichage des résultats ---
print(f"Densité 1 : {densite1:.2f} kg/m3")
print(f"Densité 2 : {densite2:.2f} kg/m3")
print(f"Densité 3 : {densite3:.2f} kg/m3")
````

### Exercice - Une question de nanosecondes

En observant les instructions 15 à 19 du code ci-dessous, que représente le nombre contenu dans la variable `mystere`?

````python
import time

def calcul_intensif():
    '''
    Cette fonction prend un certain temps pour s'exécuter.
    '''
    total = 0
    for i in range(1, 20000000):  # 20 millions d'itérations
        total += i ** 0.5
    return total

tempA = time.monotonic_ns()
calcul_intensif()
tempB = time.monotonic_ns()

mystere = tempB - tempA # Expression intéressante
print(f"La variable 'mystere' contient : {mystere}")
````

#### Solution
La variable `mystere` représente le **temps d'exécution** de la fonction `calcul_intensif()` en nanosecondes.

- `tempA` enregistre le temps juste **avant** l'appel de la fonction.
- `tempB` enregistre le temps juste **après** que la fonction a terminé son exécution.
- La différence `tempB - tempA` donne donc la durée totale écoulée pendant l'exécution de la fonction.

### Exercice - Force centripète

Écrivez un programme qui calcule la force nécessaire (Fc) pour maintenir un objet en mouvement circulaire (comme une voiture dans un virage).

**Formule :** Fc = (m ⋅ v²) / r

**Contraintes techniques :**
- Définissez une fonction nommée `calculer_force_centripete` recevant 3 paramètres : la masse, la vitesse et le rayon.
- La fonction doit retourner la valeur calculée (elle ne doit pas faire de `print` elle-même).
- Appelez la fonction, récupérez la valeur retournée dans une variable, puis affichez le résultat final avec deux décimales.

**Exemple attendu (pour m=1200, v=25, r=50) :**
```
Paramètres : 1200 kg, 25 m/s, 50 m

La force centripète résultante est de 15000.00 N
```

````python
def calculer_force_centripete(masse, vitesse, rayon):
    """Calcule la force centripète."""
    if rayon == 0:
        return float('inf')
    return (masse * vitesse**2) / rayon

# Données
m = 1200
v = 25
r = 50

# Appel de la fonction
force = calculer_force_centripete(m, v, r)

# Affichage
print(f"Paramètres : {m} kg, {v} m/s, {r} m")
print()
print(f"La force centripète résultante est de {force:.2f} N")
````
