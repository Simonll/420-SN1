### Question 3 - Écrire du code pour résoudre un problème

**Sans module ni bibliothèque.** Les relevés de température de trois stations sont stockés dans un dictionnaire :

```python
releves = {
    "Montréal": [12.5, 14.0, 9.5, 16.0],
    "Québec": [10.0, 11.5, 8.0, 12.5],
    "Gaspé": [7.5, 9.0, 6.0, 8.5],
}
```

Écrivez une fonction `analyser(releves)` qui retourne un **tuple** `(station_plus_chaude, moyennes)` où :

- `moyennes` est un dictionnaire `station -> température moyenne` ;
- `station_plus_chaude` est le nom de la station ayant la moyenne la plus élevée.

N'utilisez ni `sum()`, ni `max()`.

````python
def analyser(releves):
    moyennes = {}
    for station, temperatures in releves.items():
        total = 0
        for t in temperatures:
            total += t
        moyennes[station] = total / len(temperatures)

    station_plus_chaude = None
    for station, moyenne in moyennes.items():
        if station_plus_chaude is None or moyenne > moyennes[station_plus_chaude]:
            station_plus_chaude = station

    return station_plus_chaude, moyennes


releves = {
    "Montréal": [12.5, 14.0, 9.5, 16.0],
    "Québec": [10.0, 11.5, 8.0, 12.5],
    "Gaspé": [7.5, 9.0, 6.0, 8.5],
}
plus_chaude, moyennes = analyser(releves)
print("Moyennes :", moyennes)
print("Station la plus chaude :", plus_chaude)
````

### Question 4 - Graphique sans Pandas (données à adapter)

Une station météo a enregistré ses données sous forme de texte, une ligne par mois au format `mois;précipitations_mm;température_C` :

```python
donnees_brutes = """jan;78;-9.7
fév;62;-8.1
mar;71;-2.2
avr;83;6.0
mai;81;13.1
jun;87;18.4"""
```

1. Transformez ces données en trois listes : `mois`, `precipitations` (nombres) et `temperatures` (nombres) ;
2. Tracez les précipitations sous forme de **diagramme à barres** ;
3. Sur un **second graphique**, tracez la température en fonction du mois (courbe avec marqueurs) et une ligne horizontale à 0 °C ;
4. Ajoutez titres, labels d'axes et légende ; sauvegardez le second graphique dans `temperatures.png`.

````python
donnees_brutes = """jan;78;-9.7
fév;62;-8.1
mar;71;-2.2
avr;83;6.0
mai;81;13.1
jun;87;18.4"""

mois = []
precipitations = []
temperatures = []
for ligne in donnees_brutes.split("\n"):
    nom, precip, temp = ligne.strip().split(";")
    mois.append(nom)
    precipitations.append(int(precip))
    temperatures.append(float(temp))

plt.bar(mois, precipitations, color="steelblue", label="Précipitations")
plt.title("Précipitations mensuelles")
plt.xlabel("Mois")
plt.ylabel("Précipitations (mm)")
plt.legend()
plt.show()

plt.plot(mois, temperatures, "o-", color="darkorange", label="Température moyenne")
plt.axhline(0, color="gray", linestyle="--", label="0 °C")
plt.title("Température mensuelle moyenne")
plt.xlabel("Mois")
plt.ylabel("Température (°C)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig("temperatures.png")
plt.show()
````

### Question 5 - Analyse d'un fichier CSV avec Pandas

Utilisez le fichier `mpg_dataset.csv` et répondez aux questions suivantes :

1. Combien de voitures le fichier contient-il, et combien de valeurs manquantes y a-t-il par colonne ?
2. Combien de voitures y a-t-il par région d'origine (`origin`) ?
3. Quelle est la consommation moyenne (`mpg`) selon le nombre de cylindres ?
4. Quelles sont les 5 voitures japonaises les plus lourdes (`name`, `weight`) ?
5. Parmi les voitures de 1980 et plus (`model_year >= 80`), combien font plus de 30 mpg ?
6. Après avoir retiré les lignes où `horsepower` est manquant, quelle est la puissance moyenne par origine ?

````python
import pandas as pd

df = pd.read_csv(chemin_donnees("mpg_dataset.csv"))

print("1. Nombre de voitures :", len(df))
print(df.isna().sum())

print("\n2. Voitures par origine :")
print(df.groupby("origin")["name"].count())

print("\n3. mpg moyen par nombre de cylindres :")
print(df.groupby("cylinders")["mpg"].mean().round(1))

print("\n4. Les 5 voitures japonaises les plus lourdes :")
japonaises = df[df["origin"] == "japan"]
print(japonaises.sort_values("weight", ascending=False).head(5)[["name", "weight"]])

recentes_econo = df[(df["model_year"] >= 80) & (df["mpg"] > 30)]
print("\n5. Voitures de 1980+ à plus de 30 mpg :", len(recentes_econo))

df_hp = df.dropna(subset=["horsepower"])
print("\n6. Puissance moyenne par origine :")
print(df_hp.groupby("origin")["horsepower"].mean().round(1))
````

### Question 6 - Synthèse

Le restaurant du fichier `tips_dataset.csv` veut un petit rapport.

1. Chargez les données et ajoutez une colonne `pourcentage` = `tip / total_bill * 100` ;
2. Avec `map()`, ajoutez une colonne `jour` en français (`Thur` -> `Jeudi`, `Fri` -> `Vendredi`, `Sat` -> `Samedi`, `Sun` -> `Dimanche`) ;
3. Avec `apply()`, ajoutez une colonne `categorie` : « excellent » si le pourcentage dépasse 20, « correct » s'il est entre 15 et 20, « faible » sinon ;
4. Calculez, par jour, le pourcentage moyen de pourboire et le nombre de repas ;
5. Exportez ce résumé dans `resume_pourboires.json` avec `to_json()`, puis relisez-le avec `pd.read_json()` ;
6. Tracez avec Matplotlib (sans les graphiques de Pandas) un diagramme à barres du pourcentage moyen par jour, dans l'ordre Jeudi, Vendredi, Samedi, Dimanche.

````python
tips = pd.read_csv(chemin_donnees("tips_dataset.csv"))
tips["pourcentage"] = tips["tip"] / tips["total_bill"] * 100

traduction = {"Thur": "Jeudi", "Fri": "Vendredi", "Sat": "Samedi", "Sun": "Dimanche"}
tips["jour"] = tips["day"].map(traduction)


def categoriser(pourcentage):
    if pourcentage > 20:
        return "excellent"
    elif pourcentage >= 15:
        return "correct"
    else:
        return "faible"


tips["categorie"] = tips["pourcentage"].apply(categoriser)
print(tips[["total_bill", "tip", "pourcentage", "jour", "categorie"]].head())
print("\nRépartition des catégories :")
print(tips.groupby("categorie")["tip"].count())

resume = pd.DataFrame({
    "pourcentage_moyen": tips.groupby("jour")["pourcentage"].mean().round(2),
    "nb_repas": tips.groupby("jour")["tip"].count(),
})
print("\nRésumé par jour :")
print(resume)

resume.to_json("resume_pourboires.json")
print("\nRelu depuis le JSON :")
print(pd.read_json("resume_pourboires.json"))

ordre = ["Jeudi", "Vendredi", "Samedi", "Dimanche"]
valeurs = [resume.loc[j, "pourcentage_moyen"] for j in ordre]
plt.bar(ordre, valeurs, color="seagreen")
plt.title("Pourcentage moyen de pourboire par jour")
plt.xlabel("Jour")
plt.ylabel("Pourboire moyen (% de la facture)")
plt.show()
````
