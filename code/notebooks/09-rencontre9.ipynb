### Exercice 1 - Ensemble de données (set)

Écrivez un programme qui utilise des ensembles pour :

1. Créer deux ensembles avec des données de votre choix ;
2. Afficher leur union, leur intersection et leur différence ;
3. Vérifier si un élément est présent dans l'un ou l'autre des ensembles.

**Rappel :** un ensemble ne contient jamais de doublons et n'a pas d'ordre.

````python
cours_physique = {"Alice", "Bob", "Chloé", "David", "Émile"}
cours_chimie = {"Chloé", "Émile", "Fatima", "Gabriel"}

print("Union        :", cours_physique | cours_chimie)
print("Intersection :", cours_physique & cours_chimie)
print("Différence   :", cours_physique - cours_chimie)

nom = "Fatima"
if nom in cours_physique or nom in cours_chimie:
    print(f"{nom} est inscrit(e) dans au moins un des deux cours.")
else:
    print(f"{nom} n'est inscrit(e) dans aucun des deux cours.")

# Les méthodes équivalentes aux opérateurs
print(cours_physique.union(cours_chimie) == cours_physique | cours_chimie)
````

### Exercice 2 - Tuple de données

Un produit dans un magasin est représenté par un tuple contenant :

- le nom du produit (`str`) ;
- le prix (`float`) ;
- la quantité en stock (`int`).

```python
produit = ("Laptop", 999.99, 50)
```

Écrivez un programme qui :

1. Affiche les informations du produit de manière lisible ;
2. Calcule la valeur totale du stock ;
3. Tente (et intercepte) une modification du tuple pour montrer son immuabilité.

````python
produit = ("Laptop", 999.99, 50)

# Déballage (unpacking) du tuple
nom, prix, quantite = produit

print(f"Produit  : {nom}")
print(f"Prix     : {prix:.2f} $")
print(f"En stock : {quantite} unités")

valeur_stock = prix * quantite
print(f"Valeur totale du stock : {valeur_stock:.2f} $")

try:
    produit[1] = 899.99
except TypeError as erreur:
    print("Modification impossible :", erreur)

# Pour « modifier » un tuple, on en crée un nouveau
produit_solde = (nom, 899.99, quantite)
print("Nouveau tuple :", produit_solde)
````

### Exercice 3 - Dictionnaire de contacts

Créez un dictionnaire qui représente des contacts :

```python
contacts = {
    "alice": {"email": "alice@example.com", "telephone": "555-0101", "entreprise": "TechCorp"},
    "bob": {"email": "bob@example.com", "telephone": "555-0102", "entreprise": "DataInc"},
    "charlie": {"email": "charlie@example.com", "telephone": "555-0103", "entreprise": "AnalyticsLLC"}
}
```

Écrivez un programme qui :

1. Affiche tous les contacts avec leurs informations ;
2. Permet de rechercher un contact par nom ;
3. Ajoute un nouveau contact ;
4. Supprime un contact (cela doit fonctionner même si le contact n'existe pas : gérez l'erreur).

````python
contacts = {
    "alice": {"email": "alice@example.com", "telephone": "555-0101", "entreprise": "TechCorp"},
    "bob": {"email": "bob@example.com", "telephone": "555-0102", "entreprise": "DataInc"},
    "charlie": {"email": "charlie@example.com", "telephone": "555-0103", "entreprise": "AnalyticsLLC"}
}


def afficher_contacts(contacts):
    for nom, infos in contacts.items():
        print(f"{nom.capitalize():<10} | {infos['email']:<22} | {infos['telephone']} | {infos['entreprise']}")


def rechercher_contact(contacts, nom):
    nom = nom.lower()
    if nom in contacts:
        return contacts[nom]
    return None


def supprimer_contact(contacts, nom):
    try:
        del contacts[nom.lower()]
        print(f"Contact « {nom} » supprimé.")
    except KeyError:
        print(f"Le contact « {nom} » n'existe pas : rien à supprimer.")


print("1. Tous les contacts :")
afficher_contacts(contacts)

print("\n2. Recherche de « Bob » :", rechercher_contact(contacts, "Bob"))
print("   Recherche de « Zoé » :", rechercher_contact(contacts, "Zoé"))

contacts["diane"] = {"email": "diane@example.com", "telephone": "555-0104", "entreprise": "BioLab"}
print("\n3. Après l'ajout de Diane :")
afficher_contacts(contacts)

print("\n4. Suppressions :")
supprimer_contact(contacts, "bob")
supprimer_contact(contacts, "bob")
afficher_contacts(contacts)
````

### Exercice 4 - Compréhension de liste avancée

Écrivez un programme qui utilise la compréhension de liste pour :

1. Générer une liste de 10 nombres aléatoires entre 1 et 100 ;
2. Créer une nouvelle liste ne contenant que les nombres pairs ;
3. Créer une nouvelle liste avec les carrés des nombres impairs ;
4. Créer une liste de chaînes formatées : « Le nombre X est pair/impair ».

````python
import random

random.seed(9)  # pour obtenir les mêmes nombres à chaque exécution

nombres = [random.randint(1, 100) for _ in range(10)]
pairs = [n for n in nombres if n % 2 == 0]
carres_impairs = [n ** 2 for n in nombres if n % 2 != 0]
descriptions = [f"Le nombre {n} est {'pair' if n % 2 == 0 else 'impair'}" for n in nombres]

print("Nombres          :", nombres)
print("Pairs            :", pairs)
print("Carrés impairs   :", carres_impairs)
for phrase in descriptions:
    print(phrase)
````

### Exercice 5 - Manipulation de fichiers texte

1. Créez un fichier texte `population.txt` avec le mode `"w"` contenant des données fictives (noms de villes et leur population, une ville par ligne) ;
2. Réouvrez le fichier en mode `"r"` et comptez le nombre de lignes ;
3. Réouvrez le fichier et créez une liste contenant uniquement les noms des villes (sans la population) en utilisant une boucle `for ligne in f` et `split()` ;
4. Ajoutez une nouvelle ville au fichier avec le mode `"a"`.

**Rappel :** le mode `"w"` écrase le contenu existant, alors que le mode `"a"` ajoute à la fin du fichier.

````python
# 1. Écriture (le mode "w" crée le fichier ou l'écrase)
with open("population.txt", "w", encoding="utf-8") as f:
    f.write("Montréal;1762949\n")
    f.write("Québec;549459\n")
    f.write("Laval;438366\n")
    f.write("Gatineau;291041\n")

# 2. Compter les lignes
with open("population.txt", "r", encoding="utf-8") as f:
    lignes = f.readlines()
print("Nombre de lignes :", len(lignes))

# 3. Extraire les noms des villes
villes = []
with open("population.txt", "r", encoding="utf-8") as f:
    for ligne in f:
        nom_ville, population = ligne.strip().split(";")
        villes.append(nom_ville)
print("Villes :", villes)

# 4. Ajout d'une ville (le mode "a" ajoute à la fin)
with open("population.txt", "a", encoding="utf-8") as f:
    f.write("Sherbrooke;172950\n")

with open("population.txt", "r", encoding="utf-8") as f:
    print("\nContenu final du fichier :")
    print(f.read())
````

### Exercice 7 - Sérialisation en JSON (écriture et lecture)

Jusqu'ici vous avez lu des données ; maintenant vous allez les **écrire** dans un fichier JSON.

1. Créez un dictionnaire `etudiants` contenant 3 entrées (clé = nom, valeur = dictionnaire avec `age` et `programme`) ;
2. Ouvrez (ou créez) un fichier `etudiants.json` en mode `"w"` avec `open()` ;
3. Utilisez `json.dump(etudiants, f, ensure_ascii=False, indent=2)` pour y écrire le dictionnaire (accents préservés, indentation lisible) ;
4. Rouvrez le fichier en mode `"r"` et utilisez `json.load(f)` pour le relire et afficher son contenu.

````python
import json

etudiants = {
    "Alice": {"age": 19, "programme": "200.B1"},
    "Bob": {"age": 20, "programme": "200.BC"},
    "Éloïse": {"age": 18, "programme": "200.BA"},
}

with open("etudiants.json", "w", encoding="utf-8") as f:
    json.dump(etudiants, f, ensure_ascii=False, indent=2)

with open("etudiants.json", "r", encoding="utf-8") as f:
    etudiants_relus = json.load(f)

for nom, infos in etudiants_relus.items():
    print(f"{nom} : {infos['age']} ans, programme {infos['programme']}")

print("\nLa sauvegarde est identique à l'original :", etudiants == etudiants_relus)
````

### Exercice 8 - Lecture d'un JSON avec Pandas

Pandas peut aussi lire un fichier JSON et le transformer directement en DataFrame.

1. Écrivez dans un fichier `eleves.json` une **liste** de dictionnaires (un dictionnaire par élève, avec `nom`, `age` et `note`) ;
2. Lisez ce fichier avec `pd.read_json()` et affichez le DataFrame ;
3. Affichez la note moyenne des élèves.

````python
eleves = [
    {"nom": "Alice", "age": 20, "note": 85},
    {"nom": "Bob", "age": 22, "note": 90},
    {"nom": "Charlie", "age": 19, "note": 78},
]

with open("eleves.json", "w", encoding="utf-8") as f:
    json.dump(eleves, f, ensure_ascii=False, indent=2)

df_eleves = pd.read_json("eleves.json")
print(df_eleves)
print("\nNote moyenne :", df_eleves["note"].mean())
````
