## 3. Expressions

> **Idée importante :** une expression produit toujours une valeur.

En **mathématiques**, une expression est quelque chose qu'on peut **calculer** :
- des nombres, des constantes (π), des variables (r)
- des opérations (+, -, ×, /)

Exemple : $2\pi r$ (circonférence d'un cercle)

En **programmation**, c'est le même principe : une expression est quelque chose que l'ordinateur **évalue** pour produire une valeur. La différence : une expression peut contenir autre chose que des nombres (texte, fonctions, booléens...).

Quelques expressions Python :

````python
5          # → 5 (type : int)
"5"        # → "5" (type : str)
5 + 3      # → 8
"5" + "3" # → "53" (concaténation, pas addition!)
````

> **Même opérateur `+`, résultat différent selon le type** (surcharge de l'opérateur).

## 4. Opérateurs

### Opérateurs arithmétiques

````python
# Opérateurs arithmétiques de base
10 + 3   # Addition  → 13
10 - 3   # Soustraction → 7
10 * 3   # Multiplication → 30
11 / 3   # Division (float) → 3.6666...
11 // 3  # Division entière → 3 (partie décimale ignorée)
11 % 3   # Modulo (reste) → 2
10 ** 3  # Puissance → 1000
````

### Modulo `%`

Le modulo trouve le **reste** d'une division entière.

Exemple : 17 modulo 5
1. Combien de fois 5 entre-t-il entièrement dans 17 ? → 3 fois (3 × 5 = 15)
2. Différence : 17 − 15 = 2
3. Reste = 2 → `17 % 5 == 2`

Utilité : tester si un nombre est pair (`n % 2 == 0`) ou impair (`n % 2 == 1`).

````python
# Modulo — pair ou impair
0 % 2  # 0 → pair
1 % 2  # 1 → impair
2 % 2  # 0 → pair
3 % 2  # 1 → impair
17 % 5 # 2
````

### Division `/` vs `//`

````python
11 / 3   # → 3.6666... (float)
11 // 3  # → 3 (int, partie décimale ignorée, pas d'arrondi)
4 / 2    # → 2.0 (float, même si résultat entier)
4 // 2   # → 2 (int)
````

### Priorité des opérateurs

Certains opérateurs sont évalués avant d'autres.
- `10 + 3 * 2` → multiplication d'abord → **16**
- `(10 + 3) * 2` → parenthèses forcent l'addition en premier → **26**

Les parenthèses `()` regroupent des expressions pour forcer l'ordre d'évaluation.

````python
10 + 3 * 2    # → 16 (multiplication d'abord)
(10 + 3) * 2  # → 26 (parenthèses d'abord)
````

### Opérateurs d'affectation

````python
x = 5   # affectation simple : stocke 5 dans x
x       # affiche le contenu de x
````

> `=` n'est pas un symbole d'égalité (comme en maths), mais un symbole d'**affectation** : on calcule d'abord l'expression à droite, puis on stocke le résultat dans la variable à gauche.

## 5. Types

En programmation, les valeurs ont un **type** qui détermine comment on peut les utiliser.

| Type | Description | Exemple |
|------|-------------|---------|
| `int` | Nombre entier | `42` |
| `float` | Nombre décimal | `3.14` |
| `str` | Chaîne de caractères | `"Alice"` |
| `bool` | Booléen (vrai/faux) | `True` |

````python
# Exemples de types
42        # int
3.14      # float
"Alice"  # str
True      # bool
````

### `type()` — connaître le type d'une expression

````python
type(42)        # <class 'int'>
type('42')      # <class 'str'> — une chaîne, pas un nombre!
type(6 * 7)    # <class 'int'>
type(3.14)     # <class 'float'>
type("Alice")  # <class 'str'>
type(True)      # <class 'bool'>
type('False')  # <class 'str'> — entre guillemets, c'est une chaîne
````

### Changement de type (conversion)

````python
str(42)        # "42" — int → str
int("42")     # 42 — str → int
float(3)       # 3.0 — int → float
float("3.14") # 3.14 — str → float
````

### Polymorphisme des opérateurs

Certains opérateurs se comportent différemment selon le type de leurs opérandes.

L'opérateur `+` :

````python
5 + 3                # → 8 (addition d'entiers)
"coucou " + "Alice" # → "coucou Alice" (concaténation de chaînes)
"coucou " + str(81) # → "coucou 81" (chaîne + entier converti)
````

> **Attention :** `"123"` est une chaîne, pas un nombre entier! Pour un nombre, écrire `123` sans guillemets.
> `True` est un booléen, pas une chaîne! Pour une chaîne, écrire `"True"`.

## 6. Variables

Les variables stockent des informations manipulées par l'application, temporairement en mémoire vive (RAM).

### Syntaxe pour créer une variable

```
nom_de_ma_variable = expression
```

- `nom_de_ma_variable` : nom unique de la variable
- `=` : opérateur d'affectation (stocke la valeur de l'expression dans la variable)
- `expression` : une valeur, une opération, ou une chaîne

L'expression est d'abord évaluée, puis la valeur résultante est stockée dans la variable.

> `=` a une priorité plus basse que les opérateurs arithmétiques : l'expression à droite est évaluée en premier.

````python
x = 5 * 6    # (5 * 6) est évalué → 30, puis stocké dans x
y = 2 * x    # (2 * x) est évalué → 60, puis stocké dans y
x            # affiche 30
y            # affiche 60
````

### Voir le contenu d'une variable

Tapez le nom de la variable dans la console et appuyez sur `Entrée`.

````python
x = 42
x   # affiche 42
````

### Règles pour nommer une variable

- Lettres, chiffres (0-9) et underscores `_` uniquement.
- Ne peut pas commencer par un chiffre.
- Pas d'espaces ni de caractères spéciaux (`!`, `@`, `#`, etc.).
- Ne pas utiliser les mots réservés de Python (`if`, `else`, `while`, `for`, etc.).
- Accents déconseillés (Python les accepte, mais pas convention).

**Convention :**
- Minuscules pour les variables dont le contenu peut évoluer.
- Majuscules pour les constantes immuables (ex: `PI`).
- `snake_case` (underscores) pour les noms composés.

````python
rayon_cercle = 20                    # int
PI = 3.1416                          # float (constante)
nom_element_chimique = "Hydrogène"   # str
est_halogene = True                  # bool
circonference_circle = rayon_cercle * PI  # float

print("*** VALEURS DES VARIABLES ***")
print("rayon_cercle :", rayon_cercle)
print("PI :", PI)
print("nom_element_chimique :", nom_element_chimique)
print("est_halogene :", est_halogene)
print("circonference_circle :", circonference_circle)
````

### Les variables peuvent varier

Le contenu d'une variable peut être modifié au fur et à mesure de l'exécution.

````python
nombre1 = 10
nombre2 = 20
nombre3 = 30
resultat = nombre1 + nombre2   # 10 + 20 = 30
resultat = resultat + nombre3  # 30 + 30 = 60
resultat = resultat / 3        # 60 / 3 = 20.0
print("resultat :", resultat)  # → 20.0
````
