### Exercice - Jeu de devinette numérique

Écrivez un programme qui choisit un nombre entier aléatoire entre 0 et 100. L'utilisateur doit deviner ce nombre.

Tant que la réponse est incorrecte, le programme indique si le nombre à deviner est "Plus haut!" ou "Plus bas!".
Lorsqu'il trouve le bon nombre, le programme le félicite et s'arrête.

````python
import random

nombre_secret = random.randint(0, 100)
devine = -1 # Initialisation avec une valeur qui ne peut pas être la bonne

print("J'ai choisi un nombre entre 0 et 100. À vous de deviner!")

while devine != nombre_secret:
    devine = int(input("Votre proposition : "))
    
    if devine < nombre_secret:
        print("Plus haut!")
    elif devine > nombre_secret:
        print("Plus bas!")

print(f"Bravo! Le nombre était bien {nombre_secret}.")
````

### Exercice - Umbrella Corporation, version boucle

Reprenez l'exercice de la semaine 3 sur Umbrella Corporation. Au lieu d'appeler la fonction `afficher_echantillon` trois fois manuellement, utilisez une boucle `for` et la fonction `range()` pour parcourir les listes par index.

````python
# --- Listes parallèles contenant les données ---
noms = ["Virus-T", "Virus-G", "Uroboros"]
temperatures = [37.0, 39.5, 40.0]
pH_valeurs = [6.2, 5.8, 7.0]
mutations = ["Stable", "Instable", "Agressive"]
specifiques = ["Contagiosité : Élevée", "Capacité Régénérative : Extrême", "Compatibilité Hôte : Faible"]

# --- Définition de la fonction d'affichage (inchangée) ---
def afficher_echantillon(nom, temp, ph, mutation, specifique):
    print(f"Échantillon : {nom}")
    print(f"Température : {temp}°C")
    print(f"pH : {ph}")
    print(f"Mutation Cellulaire : {mutation}")
    print(specifique)
    print("---------------")

# --- Script principal avec une boucle ---
print("Rapport - Résultats des expériences Umbrella Corporation")

for i in range(len(noms)):
    afficher_echantillon(noms[i], temperatures[i], pH_valeurs[i], mutations[i], specifiques[i])

print("Fin du rapport. Données classifiées - Niveau d'accès : Alpha")
````

### Exercice - Calculer des statistiques (sans fonctions natives)

Soit une liste de températures. Calculez et affichez le nombre de mesures, la somme, la moyenne, le minimum et le maximum **sans utiliser** les fonctions `len()`, `sum()`, `min()` et `max()`.

Vous devrez utiliser une boucle et des variables accumulatrices.

````python
temperatures = [23.5, 25.4, 27.6, 21.9, 26.7, 24.6]

if not temperatures: # Gérer le cas d'une liste vide
    print("Aucune donnée à analyser.")
else:
    # Initialisation des accumulateurs
    nb_mesures = 0
    somme = 0
    # Pour min/max, on initialise avec la première valeur de la liste
    temp_min = temperatures[0]
    temp_max = temperatures[0]

    # Boucle pour parcourir toutes les températures
    for temp in temperatures:
        nb_mesures += 1
        somme += temp
        
        if temp < temp_min:
            temp_min = temp
        if temp > temp_max:
            temp_max = temp
            
    moyenne = somme / nb_mesures

    # Affichage des résultats
    print(f"Températures : {temperatures}")
    print(f"Nombre de mesures : {nb_mesures}")
    print(f"Somme : {somme}")
    print(f"Moyenne : {moyenne:.2f}")
    print(f"Minimum : {temp_min}")
    print(f"Maximum : {temp_max}")
````

### Exercice - Affichage d'étoiles (pyramide)

Écrivez une fonction `pyramide(hauteur)` qui affiche une pyramide d'étoiles de la hauteur donnée.

**Exemple pour `pyramide(4)` :**
```
   *
  ***
 *****
*******
```
**Indices :**
- Vous aurez besoin de boucles imbriquées.
- La boucle extérieure gère les lignes (de 1 à `hauteur`).
- La boucle intérieure (ou une multiplication de chaînes) gère l'affichage des espaces et des étoiles pour chaque ligne.
- Pour une ligne `i`, il faut `hauteur - i` espaces et `2 * i - 1` étoiles.

````python
def pyramide(hauteur):
    """Affiche une pyramide d'étoiles de la hauteur spécifiée."""
    for i in range(1, hauteur + 1):
        # Calculer le nombre d'espaces et d'étoiles
        espaces = hauteur - i
        etoiles = 2 * i - 1
        
        # Construire la ligne et l'afficher
        ligne = ' ' * espaces + '*' * etoiles
        print(ligne)

# Test
pyramide(4)
````
