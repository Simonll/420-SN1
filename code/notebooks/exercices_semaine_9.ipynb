### Exercice - Graphique simple avec Matplotlib\n\nCréez un graphique linéaire simple représentant la fonction y = x² pour x allant de -5 à 5.\n\n**Instructions :**\n1. Importez `matplotlib.pyplot`.\n2. Créez une liste `x` avec des valeurs de -5 à 5.\n3. Créez une liste `y` où chaque élément est le carré de l'élément correspondant dans `x`.\n4. Tracez le graphique.\n5. Ajoutez un titre, des étiquettes pour les axes X et Y, et une grille.\n6. Affichez le graphique.

```python
import matplotlib.pyplot as plt

x = list(range(-5, 6))
y = [val**2 for val in x]

plt.plot(x, y, marker='o')
plt.title("Fonction y = x²")
plt.xlabel("Axe des x")
plt.ylabel("Axe des y")
plt.grid(True)
plt.show()
```

### Exercice - Opérations sur les ensembles (set)\n\nSoit deux listes d'étudiants inscrits à deux cours différents. Utilisez les ensembles pour trouver :\n- Les étudiants inscrits aux deux cours (intersection).\n- Tous les étudiants uniques (union).\n- Les étudiants inscrits seulement en physique (différence).

```python
physique = ["Alice", "Bob", "Charlie", "David"]
chimie = ["Charlie", "David", "Eve", "Frank"]

# Conversion en ensembles
set_physique = set(physique)
set_chimie = set(chimie)

# Intersection
intersection = set_physique & set_chimie
print(f"Étudiants inscrits aux deux cours : {intersection}")

# Union
union = set_physique | set_chimie
print(f"Tous les étudiants uniques : {union}")

# Différence
seulement_physique = set_physique - set_chimie
print(f"Étudiants seulement en physique : {seulement_physique}")
```

### Exercice - Dictionnaire d'informations\n\nCréez un dictionnaire pour stocker des informations sur un film. Affichez ensuite ces informations de manière formatée.

```python
film = {
    "titre": "Inception",
    "realisateur": "Christopher Nolan",
    "annee": 2010,
    "acteurs": ["Leonardo DiCaprio", "Joseph Gordon-Levitt", "Elliot Page"]
}

print(f"Titre : {film['titre']}")
print(f"Année de sortie : {film['annee']}")
print(f"Réalisateur : {film['realisateur']}")

# Parcourir la liste d'acteurs
print("Acteurs principaux :")
for acteur in film['acteurs']:
    print(f"- {acteur}")
```
