### Exercice - Création et exploration d'un DataFrame\n\nCréez un DataFrame Pandas à partir d'un dictionnaire et utilisez les fonctions de base pour l'explorer.

```python
import pandas as pd

data = {
    'Nom': ['Mercure', 'Vénus', 'Terre', 'Mars'],
    'Type': ['Tellurique', 'Tellurique', 'Tellurique', 'Tellurique'],
    'Diametre_km': [4879, 12104, 12742, 6779]
}

df_planetes = pd.DataFrame(data)

print("--- Aperçu du DataFrame ---")
print(df_planetes.head())

print("\n--- Informations sur le DataFrame ---")
df_planetes.info()
```

### Exercice - Filtrage et tri\n\nÀ partir du DataFrame de l'exercice précédent, filtrez les planètes ayant un diamètre supérieur à 10 000 km et triez-les par diamètre décroissant.

```python
# (On suppose que df_planetes de l'exercice précédent est disponible)

# Filtrage
grandes_planetes = df_planetes[df_planetes['Diametre_km'] > 10000]

# Tri
planetes_triees = grandes_planetes.sort_values(by='Diametre_km', ascending=False)

print("Planètes avec un diamètre > 10 000 km, triées:")
print(planetes_triees)
```
