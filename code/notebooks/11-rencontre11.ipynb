### Exercice 1 - Création et inspection d'un tableau

1. Créez un tableau NumPy contenant les nombres de 0 à 9 ;
2. Affichez sa forme (`shape`), son type de données (`dtype`), son nombre de dimensions (`ndim`) et sa taille (`size`) ;
3. Créez un second tableau 2D de forme (3, 4) rempli de zéros ;
4. Affichez sa forme et son type.

```python
import numpy as np

a = np.arange(10)
print("a     :", a)
print("shape :", a.shape)
print("dtype :", a.dtype)
print("ndim  :", a.ndim)
print("size  :", a.size)

z = np.zeros((3, 4))
print("\nz :")
print(z)
print("shape :", z.shape, "| dtype :", z.dtype)
```

### Exercice 2 - Indexation et slicing

1. Créez un tableau 2D de forme (4, 5) contenant les nombres de 0 à 19 (utilisez `arange` et `reshape`) ;
2. Affichez l'élément à la position (2, 3) ;
3. Affichez la première ligne, puis la dernière colonne ;
4. Affichez le sous-tableau formé par les 2 premières lignes et les 3 premières colonnes ;
5. Remplacez la valeur à la position (1, 2) par 99 et affichez le tableau modifié.

```python
m = np.arange(20).reshape(4, 5)
print(m)

print("\nÉlément (2, 3)     :", m[2, 3])
print("Première ligne     :", m[0])
print("Dernière colonne   :", m[:, -1])
print("Sous-tableau 2 x 3 :")
print(m[:2, :3])

m[1, 2] = 99
print("\nTableau modifié :")
print(m)
```

### Exercice 3 - Opérations vectorisées et masques booléens

1. Créez `a = [1, 2, 3, 4, 5]` et `b = [10, 20, 30, 40, 50]` sous forme de tableaux NumPy ;
2. Calculez et affichez `a + b`, `a * b` et `a ** 2` (sans boucle !) ;
3. Affichez la moyenne de `a` et le maximum de `b` ;
4. Créez un masque booléen pour les éléments de `a` supérieurs à 3, puis appliquez-le.

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print("a + b  :", a + b)
print("a * b  :", a * b)
print("a ** 2 :", a ** 2)
print("Moyenne de a :", a.mean())
print("Maximum de b :", b.max())

masque = a > 3
print("Masque       :", masque)
print("a[masque]    :", a[masque])
```

### Exercice 4 - Courbe sinus avec NumPy et Matplotlib

1. Créez un tableau `x` de 100 points équidistants entre 0 et 2π (`np.linspace`) ;
2. Calculez `y = sin(x)` ;
3. Tracez `y` en fonction de `x` ;
4. Ajoutez un titre, des labels aux axes et une grille ;
5. Sauvegardez le graphique dans `courbe_sinus.png`.

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)")
plt.title("Fonction sinus sur [0, 2π]")
plt.xlabel("x (radians)")
plt.ylabel("sin(x)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.savefig("courbe_sinus.png")
plt.show()
```

### Exercice 5 - Exploration d'un jeu de données avec NumPy

Utilisez le fichier `mpg_dataset.csv` (consommation de voitures).

1. Lisez le fichier CSV avec Pandas ;
2. Convertissez la colonne `mpg` en tableau NumPy avec `.to_numpy()` ;
3. Affichez les statistiques de base du tableau (moyenne, min, max, écart-type) ;
4. Créez un masque pour les valeurs supérieures à la moyenne ;
5. Affichez le nombre de valeurs supérieures à la moyenne.

```python
import pandas as pd

df_mpg = pd.read_csv(chemin_donnees("mpg_dataset.csv"))
mpg = df_mpg["mpg"].to_numpy()

print("Type      :", type(mpg))
print("Moyenne   :", round(mpg.mean(), 2))
print("Minimum   :", mpg.min())
print("Maximum   :", mpg.max())
print("Écart-type:", round(mpg.std(), 2))

masque_sup = mpg > mpg.mean()
print("\nNombre de voitures au-dessus de la moyenne :", masque_sup.sum(), "sur", mpg.size)
print("Exemples :", mpg[masque_sup][:10])
```

### Exercice 6 - Visualisation de données avec NumPy et Matplotlib

Utilisez le fichier `tips_dataset.csv`.

1. Chargez les données avec Pandas ;
2. Extrayez les colonnes `total_bill` et `tip` en tableaux NumPy ;
3. Tracez un nuage de points de `tip` en fonction de `total_bill` ;
4. Ajoutez un titre, des labels et une légende si nécessaire ;
5. Sauvegardez le graphique dans `tips_scatter.png`.

**Défi :** colorez différemment les repas dont le pourboire dépasse 15 % de la facture, à l'aide d'un masque booléen.

```python
df_tips = pd.read_csv(chemin_donnees("tips_dataset.csv"))
total_bill = df_tips["total_bill"].to_numpy()
tip = df_tips["tip"].to_numpy()

genereux = tip / total_bill > 0.15

plt.scatter(total_bill[~genereux], tip[~genereux], color="gray", alpha=0.6, label="Pourboire <= 15 %")
plt.scatter(total_bill[genereux], tip[genereux], color="crimson", alpha=0.7, label="Pourboire > 15 %")
plt.title("Pourboire en fonction de la facture")
plt.xlabel("Facture ($)")
plt.ylabel("Pourboire ($)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig("tips_scatter.png")
plt.show()
```

### Exercice 7 - Histogramme de données aléatoires

1. Générez 1000 nombres aléatoires suivant une loi normale (moyenne = 0, écart-type = 1) avec NumPy ;
2. Affichez la moyenne, l'écart-type, le minimum et le maximum ;
3. Tracez un histogramme de ces données (20 bins), avec un titre et des labels ;
4. Sauvegardez le graphique dans `histogramme.png`.

```python
np.random.seed(11)  # graine : mêmes nombres à chaque exécution
valeurs = np.random.normal(0, 1, 1000)

print(f"Moyenne    : {valeurs.mean():.3f}")
print(f"Écart-type : {valeurs.std():.3f}")
print(f"Minimum    : {valeurs.min():.3f}")
print(f"Maximum    : {valeurs.max():.3f}")

plt.hist(valeurs, bins=20, color="slateblue", edgecolor="white")
plt.title("1000 tirages d'une loi normale centrée réduite")
plt.xlabel("Valeur")
plt.ylabel("Fréquence")
plt.savefig("histogramme.png")
plt.show()
```

### Exercice 8 - Comparaison NumPy vs listes Python

1. Créez une liste Python de 1 million de nombres aléatoires ;
2. Créez un tableau NumPy équivalent ;
3. Mesurez le temps pour calculer la somme avec la liste Python (`sum()`) ;
4. Mesurez le temps pour calculer la somme avec NumPy (`.sum()`) ;
5. Affichez les temps de calcul et le rapport entre les deux.

**Indice :** utilisez `time.perf_counter()` du module `time`.

```python
import random
import time

liste = [random.random() for _ in range(1_000_000)]
tableau = np.array(liste)

debut = time.perf_counter()
somme_liste = sum(liste)
duree_liste = time.perf_counter() - debut

debut = time.perf_counter()
somme_numpy = tableau.sum()
duree_numpy = time.perf_counter() - debut

print(f"Somme (liste) : {somme_liste:.2f} en {duree_liste * 1000:.2f} ms")
print(f"Somme (NumPy) : {somme_numpy:.2f} en {duree_numpy * 1000:.2f} ms")
print(f"NumPy est environ {duree_liste / duree_numpy:.0f} fois plus rapide sur cette machine.")
```
